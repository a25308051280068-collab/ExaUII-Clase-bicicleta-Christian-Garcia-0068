```python
class Bicicleta:
    # Constructor para inicializar los atributos
    def __init__(self):
        self.marca = ""
        self.rodado = ""
        self.material = ""

    # Función para capturar datos (Interactivo)
    def leerDatos(self, nombre_objeto):
        print(f"\n--- Ingresando datos para {nombre_objeto} ---")
        self.marca = input("Introduce la marca: ")
        self.rodado = input("Introduce el número de rodado: ")
        self.material = input("Introduce el material del cuadro: ")

    # Función para mostrar los datos
    def mostrarDatos(self, nombre_objeto):
        print(f"\n>>> Resultados de {nombre_objeto}:")
        print(f"Marca: {self.marca}")
        print(f"Rodado: {self.rodado}")
        print(f"Material: {self.material}")
        print("-" * 30)

# --- Creación de los Objetos ---

# 1. Instanciamos los objetos
bici1 = Bicicleta()
bici2 = Bicicleta()

# 2. Llamamos a las funciones para bici1
bici1.leerDatos("bici1")
bici1.mostrarDatos("bici1")

# 3. Llamamos a las funciones para bici2
bici2.leerDatos("bici2")
bici2.mostrarDatos("bici2")

print("Programa finalizado ")
print("Programa creado por Christian Garcia NL = 21")
```
